# 01 - Dataset Inventory & Profiling

```python
import pandas as pd
from pathlib import Path
pd.set_option('display.max_columns',None)
```

```python
DATA_FOLDER=Path(r'C:\Users\Saanvi\Downloads\Mentorship')
csv_files=list(DATA_FOLDER.glob('*.csv'))
print(len(csv_files))
```

```python
inventory=[]
for file in csv_files:
    df=pd.read_csv(file,low_memory=False)
    inventory.append([file.name,len(df),len(df.columns)])
pd.DataFrame(inventory,columns=['Dataset','Rows','Columns'])
```

```python
for file in csv_files:
    df=pd.read_csv(file,nrows=5)
    print('\n'+file.name)
    print(df.columns.tolist())
```

```python
for file in csv_files:
    df=pd.read_csv(file)
    print('\n'+file.name)
    print(df.head())
    print("\n\n")
```
