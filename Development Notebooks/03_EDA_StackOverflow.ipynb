## Helper Function For Multi-Select Columns

```python
from collections import Counter

def extract_counts(series):
    counter = Counter()
    for row in series.dropna():
        for item in str(row).split(';'):
            counter[item.strip()] += 1
    return pd.Series(counter).sort_values(ascending=False)
```

## Most Used Programming Languages

```python
lang_used = extract_counts(df['LanguageHaveWorkedWith'])
lang_used.head(20)
```

```python
lang_used.head(15).sort_values().plot(kind='barh')
plt.title('Most Used Languages')
plt.show()
```
