```python
import pandas as pd

summary = []

for year in range(2011, 2026):

    file = f'../Raw Data/StackOverflow/survey_{year}.csv'

    try:
        df = pd.read_csv(file, nrows=1, low_memory=False)

    except:
        try:
            df = pd.read_csv(
                file,
                encoding='latin1',
                nrows=1,
                low_memory=False
            )

        except:
            try:
                df = pd.read_csv(
                    file,
                    encoding='cp1252',
                    nrows=1,
                    low_memory=False
                )

            except Exception as e:
                print(f'{year} FAILED -> {e}')
                continue

    cols = df.columns.tolist()

    tech_cols = [
        c for c in cols
        if any(
            k.lower() in c.lower()
            for k in [
                'language',
                'database',
                'platform',
                'framework',
                'tool',
                'technology',
                'tech',
                'haveworked',
                'workedwith'
            ]
        )
    ]

    print('\n' + '='*60)
    print(year)
    print('='*60)
    print(tech_cols)
```
