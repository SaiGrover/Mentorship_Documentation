# 08D_A_Taxonomy_Dictionary_Builder

```python
import pandas as pd
import numpy as np
```

```python
skills = pd.read_csv('../Generated Datasets/linkedin_skill_universe_1000.csv')
print(skills.shape)
```

```python
skills['skill']=skills['skill'].astype(str).str.lower().str.strip()
```

```python
freq_col=[c for c in skills.columns if c!='skill'][0]
```

```python
skills=(skills.sort_values(freq_col,ascending=False).drop_duplicates(subset='skill').reset_index(drop=True))
```

```python
skills.head(500).to_csv('../Generated Datasets/top_500_skills.csv',index=False)
```

```python
skills.head(1000).to_csv('../Generated Datasets/top_1000_skills.csv',index=False)
```

```python
taxonomy_df=skills.copy();taxonomy_df['category']='';taxonomy_df['sub_category']='';taxonomy_df['notes']='';taxonomy_df.to_csv('../Generated Datasets/skill_taxonomy_dictionary.csv',index=False)
```

```python
skills[skills[freq_col]>=10000].to_csv('../Generated Datasets/review_priority_skills.csv',index=False)
```
