# 08C_Skill_Universe_Expansion


```
import pandas as pd
import numpy as np
job_skills=pd.read_csv('../Raw Data/job_skills.csv')
master=pd.read_csv('../Generated Datasets/master_skill_dataset_v7.csv')
print(job_skills.shape)
print(master.shape)
```

```
all_skills=[]
for skills in job_skills['job_skills'].dropna():
    for skill in str(skills).split(','):
        skill=skill.strip().lower()
        if len(skill)>1:
            all_skills.append(skill)
linkedin_skill_df=pd.DataFrame({'skill':all_skills})
```

```
linkedin_freq=(linkedin_skill_df.groupby('skill').size().reset_index(name='linkedin_frequency').sort_values('linkedin_frequency',ascending=False))
linkedin_freq.head(50)
```

```
tech_skills=pd.read_excel('../Raw Data/Technology Skills.xlsx')
tech_column=tech_skills.columns[0]
tech_set=set(tech_skills[tech_column].astype(str).str.strip().str.lower())
```

```
expanded=linkedin_freq[linkedin_freq['skill'].isin(tech_set)].copy()
expanded.shape
```

```
linkedin_freq['skill']=linkedin_freq['skill'].replace({'powerbi':'power bi'})
target_skills=['langchain','llm','rag','power bi','tableau']
recovered=linkedin_freq[linkedin_freq['skill'].isin(target_skills)]
recovered
```

```
current=master['skill'].astype(str).str.lower()
expanded_universe=set(current)
for s in expanded['skill']: expanded_universe.add(s)
for s in recovered['skill']: expanded_universe.add(s)
expanded_universe=pd.DataFrame({'skill':sorted(expanded_universe)})
```

```
print('Current Skills:',len(current.unique()))
print('Expanded Skills:',len(expanded_universe))
print('Expansion Factor:',round(len(expanded_universe)/len(current.unique()),2),'x')
```

```
expanded_universe.to_csv('../Generated Datasets/expanded_skill_master.csv',index=False)
print('Saved expanded_skill_master.csv')
```
