# 08A_Build_Skill_Demand_History_From_StackOverflow

Generate skill_demand_history.csv using multiple Stack Overflow survey years.

```python
import pandas as pd
from collections import Counter
from pathlib import Path
```

```python
FILES={
2017:'../Raw Data/StackOverflow/survey_2017.csv',
2018:'../Raw Data/StackOverflow/survey_2018.csv',
2019:'../Raw Data/StackOverflow/survey_2019.csv',
2020:'../Raw Data/StackOverflow/survey_2020.csv',
2021:'../Raw Data/StackOverflow/survey_2021.csv',
2022:'../Raw Data/StackOverflow/survey_2022.csv',
2023:'../Raw Data/StackOverflow/survey_2023.csv',
2024:'../Raw Data/StackOverflow/survey_2024.csv',
2025:'../Raw Data/StackOverflow/survey_2025.csv'
}
```

```python
TECH_COLUMNS=[
'LanguageHaveWorkedWith',
'DatabaseHaveWorkedWith',
'PlatformHaveWorkedWith',
'WebframeHaveWorkedWith',
'MiscTechHaveWorkedWith',
'ToolsTechHaveWorkedWith'
]
```

```python
rows=[]

for year,file in FILES.items():

    df=pd.read_csv(file,low_memory=False)

    counter=Counter()

    available=[c for c in TECH_COLUMNS if c in df.columns]

    for col in available:

        for value in df[col].dropna():

            for skill in str(value).split(';'):

                skill=skill.strip()

                if skill:
                    counter[skill]+=1

    for skill,count in counter.items():

        rows.append([
            skill,
            year,
            count
        ])

history=pd.DataFrame(
    rows,
    columns=['skill','year','demand']
)

history.head()
```

```python
history=history.sort_values(['skill','year'])

history.to_csv(
'../Generated Datasets/skill_demand_history.csv',
index=False
)

print(history.shape)
print('Saved skill_demand_history.csv')
```
