
# 06_Growth_Rate_Feature_Engineering.ipynb

Adds a new **growth_rate** feature to the SES framework.

Formula:

`growth_rate = future_interest - current_usage`

This captures momentum and rewards skills with stronger future demand relative to current adoption.


## Growth Rate Feature Selection

Three candidate growth metrics were evaluated:

1. growth_v1 = future_interest - current_usage
2. growth_v2 = (future_interest - current_usage) × future_interest
3. growth_v3 = future_interest / (current_usage + 0.05)

After comparing rankings, growth_v2 was selected because it balances future demand and momentum while avoiding excessive inflation of low-adoption niche skills.

```python

import pandas as pd
import numpy as np

df = pd.read_csv('../Generated Datasets/master_skill_dataset_v6.csv')

print(df.shape)
df.head()
```

```python
# Version 1
# Future Interest - Current Usage

df['growth_v1'] = (
    df['future_interest']
    - df['current_usage']
)
```

```python
# Version 2 (Best Choice)
# Momentum × Future Demand

df['growth_v2'] = (
    df['future_interest']
    - df['current_usage']
) * df['future_interest']
```

```python
# Version 3
# Future / Current Ratio

df['growth_v3'] = (
    df['future_interest']
    /
    (df['current_usage'] + 0.05)
)
```

```python

# Normalize all versions

for col in [
    'growth_v1',
    'growth_v2',
    'growth_v3'
]:

    df[col] = (
        df[col]
        - df[col].min()
    ) / (
        df[col].max()
        - df[col].min()
    )

print("\nTOP 10 - GROWTH V1")
display(
    df[
        ['skill','growth_v1']
    ].sort_values(
        'growth_v1',
        ascending=False
    ).head(10)
)

print("\nTOP 10 - GROWTH V2")
display(
    df[
        ['skill','growth_v2']
    ].sort_values(
        'growth_v2',
        ascending=False
    ).head(10)
)

print("\nTOP 10 - GROWTH V3")
display(
    df[
        ['skill','growth_v3']
    ].sort_values(
        'growth_v3',
        ascending=False
    ).head(10)
)
```

```python
df['growth_rate'] = df['growth_v2']

df[['skill','growth_rate']].sort_values('growth_rate',ascending=False).head(20)
```
